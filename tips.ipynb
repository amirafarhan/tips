```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
years=[2020,2021,2022,2023,2024,2025]
sales=[564,546,800,214,454,1000]
```

```python
plt.scatter(years,sales)
plt.title("sales over the year")
plt.xlabel("years")
plt.ylabel("sales")
plt.show()
```

```python
plt.bar(years,sales)
plt.title("sales over the year")
plt.xlabel("years")
plt.ylabel("sales")
plt.show()
```

```python
plt.plot(years,sales)
plt.title("sales over the year")
plt.xlabel("years")
plt.ylabel("sales")
plt.show()
```

```python
plt.hist(sales,5,color="green")  # histogram of sales values (was plotting the years list)
plt.title("distribution of yearly sales")
plt.xlabel("sales")
plt.ylabel("number of years")
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(years,sales)
plt.title("plot")

plt.subplot(1,2,2)
plt.bar(years,sales)
plt.title("bar")
```
